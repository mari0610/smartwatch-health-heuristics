```python
import pandas as pd

ruta = '/content/Evaluación de Usabilidad_ Aplicaciones de Salud en Smartwatches (Protocolo A)(1-14).csv'

datos = pd.read_csv(ruta, sep=';')
datos.columns = datos.columns.str.strip()

n = len(datos)

edades = pd.to_numeric(datos["Edad"], errors="coerce")

edad_min = int(edades.min())
edad_max = int(edades.max())
edad_media = edades.mean()
edad_de = edades.std()
edad_mediana = edades.median()

nivel = datos["¿Qué nivel educativo posee?"].astype(str).str.strip()
conteo_nivel = nivel.value_counts()

exp_smartwatch = datos["¿Posee experiencia en el uso de Smartwatches?"].astype(str).str.strip()
conteo_smartwatch = exp_smartwatch.value_counts()

exp_heuristicas = datos["¿Posee experiencia en pruebas de usabilidad por heurísticas?"].astype(str).str.strip()
conteo_heuristicas = exp_heuristicas.value_counts()

print(f"En la condición experimental correspondiente al conjunto de heurísticas de Jakob Nielsen participaron {n} estudiantes universitarios (n = {n}). "
      f"La edad de los participantes osciló entre {edad_min} y {edad_max} años, con una edad promedio de {edad_media:.1f} años (DE = {edad_de:.1f}) "
      f"y una mediana de {edad_mediana:.0f} años.")

print()

print("En cuanto al nivel educativo, la muestra estuvo compuesta por:")
for categoria, cantidad in conteo_nivel.items():
    porcentaje = cantidad / n * 100
    print(f"- {categoria}: {cantidad} participantes ({porcentaje:.1f}%)")

print()

print("Respecto a la experiencia previa con dispositivos smartwatch:")
for categoria, cantidad in conteo_smartwatch.items():
    porcentaje = cantidad / n * 100
    print(f"- {categoria}: {cantidad} participantes ({porcentaje:.1f}%)")

print()

print("Respecto a la experiencia previa en pruebas de usabilidad por heurísticas:")
for categoria, cantidad in conteo_heuristicas.items():
    porcentaje = cantidad / n * 100
    print(f"- {categoria}: {cantidad} participantes ({porcentaje:.1f}%)")
```

```python
import pandas as pd
from prettytable import PrettyTable

ruta = '/content/Evaluación de Usabilidad_ Aplicaciones de Salud en Smartwatches (Protocolo A)(1-14).csv'

datos = pd.read_csv(ruta, sep=';')
datos.columns = datos.columns.str.strip()

columnas_nielsen = datos.columns[10:18]

heuristicas_nielsen = ["H1", "H2", "H3", "H4", "H5", "H6", "H7", "H8"]

tabla = PrettyTable()
tabla.field_names = ["Pregunta evaluada", "Heurística de Nielsen", "0", "1", "2", "3", "4"]

for i, col in enumerate(columnas_nielsen):
    respuestas = (
        datos[col]
        .astype(str)
        .str.replace('\xa0', '', regex=False)
        .str.strip()
    )

    respuestas = pd.to_numeric(respuestas, errors="coerce")
    frecuencias = respuestas.value_counts().reindex([0, 1, 2, 3, 4], fill_value=0)

    tabla.add_row([
        f"Pregunta {i + 1}",
        heuristicas_nielsen[i],
        int(frecuencias[0]),
        int(frecuencias[1]),
        int(frecuencias[2]),
        int(frecuencias[3]),
        int(frecuencias[4])
    ])

print(tabla)
```
